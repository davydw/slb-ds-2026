## 📚 Libraries

```python
### DATA MANIPULATION
import pandas as pd
pd.set_option('display.max_columns', None)
import numpy as np

### VISUALIZATION
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

## 💾 Load the dataset

```python
data = pd.read_csv("emissions.csv")
data.head()
```

Filter this DataFrame to keep rows where the year is >= 1950.

```python
boolean_filer = data["year"] >= 1950
```

```python
data = data[boolean_filer]
```

```python
data["year"].describe()
```

```python
data.head(20)
```

```python
# Use the column 'year' as the index of your dataframe
data.set_index("year", inplace=True)
```

```python
filter_MY = data["country"] == "Malaysia"
```

```python
malaysia_data =  data[filter_MY]
malaysia_data.tail(10)
```

### 2- Matplotlib

```python
x = malaysia_data.index
x
```

```python
y = malaysia_data["gdp"]
y
```

```python
plt.figure(figsize=(10, 5))
plt.plot(x,y,
         c="red",
         linestyle='dotted',
         label="GDP"
         # marker="$s$"
         );

plt.plot(x,malaysia_data["population"],
         c="blue",
         linestyle='dashed',
         label="POP"
         # marker="$s$"
         );

plt.legend()
plt.grid()
```

```python
fig = plt.figure(figsize=(10, 3))  # Create a `Figure`, your canvas

plt.subplot(1, 2, 1)               # Start a first subplot
plt.plot(malaysia_data.index, malaysia_data['oil_co2'])
plt.title('oil_co2')

plt.subplot(1, 2, 2)               # Start a second subplot
plt.plot(malaysia_data.index, malaysia_data['gas_co2'])
plt.title('gas_co2');
```
